```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/wine_data.csv',header=None,usecols=[0,1,2])
df.columns=['Class Label','Alcohol','Malic Acid']
```

```python
df.head()
```

```python
plt.figure(figsize=(10,6))
sns.kdeplot(data=df,x='Alcohol')
```

```python
plt.figure(figsize=(10,6))
sns.kdeplot(data=df,x='Malic Acid')
```

```python
color_dict={1:'red',2:'bule',3:'green'}
sns.scatterplot(df,x=df['Alcohol'],y=df['Malic Acid'],hue=df['Class Label'].map(color_dict))
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.drop('Class Label',axis=1),df['Class Label'],test_size=0.3,random_state=0)
X_train.shape,X_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
scaler.fit(X_train)
X_train_scaled=scaler.transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
X_train_scaled=pd.DataFrame(X_train_scaled,columns=X_train.columns)
X_test_scaled=pd.DataFrame(X_test_scaled,columns=X_test.columns)
```

```python
np.round(X_train.describe(),1)
```

```python
np.round(X_train_scaled.describe(),1)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(15,6))
ax1.scatter(X_train['Alcohol'],X_train['Malic Acid'],c=y_train)
ax1.set_title('Before Scalling')
ax2.scatter(X_train_scaled['Alcohol'],X_train_scaled['Malic Acid'],c=y_train)
ax2.set_title('After Scalling')
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(15,6))
sns.kdeplot(X_train['Alcohol'],ax=ax1)
sns.kdeplot(X_train['Malic Acid'],ax=ax1)
ax1.set_title('Before Scalling')
sns.kdeplot(X_train_scaled['Alcohol'],ax=ax2)
ax2.set_title('After Scalling')
```
