```python
import numpy as np
import pandas as pd
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
import seaborn as sns
```

```python
df=pd.read_csv('train.csv',usecols=['Pclass','Age','SibSp','Parch','Survived'])
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X=df.drop(columns=['Survived'])
Y=df['Survived']
```

```python
np.mean(cross_val_score(LogisticRegression(),X,Y,scoring='accuracy',cv=10))
```

```python
X['Family_size']=X['Parch']+X['SibSp']+1
```

```python
X.head()
```

```python
def func(num):
  if num==1:
    return 0
  elif num>1 or num>=4:
    return 1
  else:
    return 2
```

```python
func(4)
```

```python
X['Family_type']=X['Family_size'].apply(func)
```

```python
X.head()
```

```python
X.drop(columns=['Parch','SibSp','Family_size'],inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,Y,scoring='accuracy',cv=10))
```
