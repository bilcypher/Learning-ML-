```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips=sns.load_dataset('tips')
titanic = pd.read_csv('/content/train.csv')
flights=sns.load_dataset('flights')
iris=sns.load_dataset('iris')
```

```python
plt.figure(figsize=(15,10))
sns.scatterplot(tips,x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'],size=tips['size'])
```

```python
titanic.head()
```

```python
sns.barplot(titanic,x=titanic['Pclass'],y=titanic['Fare'],palette='viridis',hue=titanic['Sex'])
```

```python
sns.boxplot(titanic,x=titanic['Sex'],y=titanic['Age'],palette='viridis',hue=titanic['Survived'])
```

```python
sns.displot(titanic[titanic['Survived']==0]['Age'],kind='kde',color='red')
sns.displot(titanic[titanic['Survived']==1]['Age'],kind='kde')
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
(titanic.groupby('Pclass')['Survived'].mean()).plot(kind='bar')
```

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```

```python
sns.pairplot(iris,hue='species')
```

```python
flights.head()
new=flights.groupby('year')['passengers'].sum().reset_index()
sns.lineplot(data=new,x='year',y='passengers')
```

```python
sns.heatmap(flights.pivot_table(values='passengers',index='month',columns='year'))
```
